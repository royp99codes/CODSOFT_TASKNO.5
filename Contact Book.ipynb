***CODESOFT PYTHON PROGRAMMING***


```python
class ContactBook:
    def __init__(self):
        # Using a list of dictionaries to store contacts
        self.contacts = []

    def add_contact(self):
        print("\n--- ADD NEW CONTACT ---")
        name = input("Enter Name: ").strip()
        if not name:
            print("Name cannot be empty.")
            return

        phone = input("Enter Contact No.: ").strip()
        email = input("Enter Email: ").strip()
        address = input("Enter Address: ").strip()

        # Save as a dictionary item
        contact = {
            "name": name,
            "phone": phone,
            "email": email,
            "address": address,
        }
        self.contacts.append(contact)
        print(f"Contact for '{name}' added successfully!")

    def view_contacts(self):
        print("\n--- ALL CONTACTS ---")
        if not self.contacts:
            print("No contacts found.")
            return

        print(
            f"{'Index':<6} | {'Name':<20} | {'Contact No.':<15} | {'Email':<25} | {'Address':<20}"
        )
        print("-" * 92)
        for i, contact in enumerate(self.contacts, 1):
            print(
                f"{i:<6} | {contact['name']:<20} | {contact['phone']:<15} | {contact['email']:<25} | {contact['address']:<20}"
            )

    def search_contact(self):
        """Helper method to find contacts by Name or Phone Number."""
        query = input("\nEnter Name or Phone No. to search: ").strip().lower()
        results = [
            (idx, c)
            for idx, c in enumerate(self.contacts)
            if query in c["name"].lower() or query in c["phone"].lower()
        ]
        return results

    def update_contact(self):
        print("\n--- UPDATE CONTACT ---")
        if not self.contacts:
            print("No contacts available to update.")
            return

        results = self.search_contact()
        if not results:
            print("No matching contact found.")
            return

        # Display matching results
        print("\nMatching Contacts:")
        for display_idx, (original_idx, c) in enumerate(results, 1):
            print(
                f"{display_idx}. {c['name']} | Phone: {c['phone']} | Email: {c['email']}"
            )

        try:
            choice = int(
                input(
                    "\nSelect the number of the contact you want to update: "
                )
            )
            if 1 <= choice <= len(results):
                target_idx = results[choice - 1][0]
                contact = self.contacts[target_idx]

                print("\nLeave field blank to keep current value:")
                new_name = (
                    input(f"Name [{contact['name']}]: ").strip()
                    or contact["name"]
                )
                new_phone = (
                    input(f"Phone [{contact['phone']}]: ").strip()
                    or contact["phone"]
                )
                new_email = (
                    input(f"Email [{contact['email']}]: ").strip()
                    or contact["email"]
                )
                new_address = (
                    input(f"Address [{contact['address']}]: ").strip()
                    or contact["address"]
                )

                self.contacts[target_idx] = {
                    "name": new_name,
                    "phone": new_phone,
                    "email": new_email,
                    "address": new_address,
                }
                print("Contact updated successfully!")
            else:
                print("Invalid choice.")
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    def delete_contact(self):
        print("\n--- DELETE CONTACT ---")
        if not self.contacts:
            print("No contacts available to delete.")
            return

        results = self.search_contact()
        if not results:
            print("No matching contact found.")
            return

        print("\nMatching Contacts:")
        for display_idx, (original_idx, c) in enumerate(results, 1):
            print(
                f"{display_idx}. {c['name']} | Phone: {c['phone']} | Email: {c['email']}"
            )

        try:
            choice = int(
                input(
                    "\nSelect the number of the contact you want to delete: "
                )
            )
            if 1 <= choice <= len(results):
                target_idx = results[choice - 1][0]
                removed = self.contacts.pop(target_idx)
                print(f"Contact for '{removed['name']}' deleted successfully!")
            else:
                print("Invalid choice.")
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    def run(self):
        while True:
            print("\n==================================")
            print("       CONTACT BOOK MANAGER       ")
            print("==================================")
            print("1. Add New Contact")
            print("2. View All Contacts")
            print("3. Update Contact")
            print("4. Delete Contact")
            print("5. Exit")
            print("==================================")

            choice = input("Enter your choice (1-5): ").strip()

            if choice == "1":
                self.add_contact()
            elif choice == "2":
                self.view_contacts()
            elif choice == "3":
                self.update_contact()
            elif choice == "4":
                self.delete_contact()
            elif choice == "5":
                print("\nThank you for using Contact Book Manager! Goodbye!")
                break
            else:
                print("❌ Invalid option. Please select a number between 1 and 5.")


if __name__ == "__main__":
    book = ContactBook()
    book.run()
```
